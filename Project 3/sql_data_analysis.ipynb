# Import Libraries

```python
import sqlite3
import pandas as pd
```

# Load Dataset

```python
# Load the Excel dataset into a pandas DataFrame
df = pd.read_excel('Dataset for Data Analytics.xlsx')
df.head()
```

# Create Database

```python
# Create SQLite database connection and load table
conn = sqlite3.connect('ecommerce.db')
df.to_sql('orders', conn, if_exists='replace', index=False)
```

# SELECT Queries
This section displays records from the orders table using basic SELECT statements.

```python
# This query displays all columns for the first 5 records
query1 = 'SELECT * FROM orders LIMIT 5;'
pd.read_sql_query(query1, conn)
```

```python
# This query selects specific columns (OrderID, Product, Quantity, TotalPrice) for 5 rows
query2 = 'SELECT OrderID, Product, Quantity, TotalPrice FROM orders LIMIT 5;'
pd.read_sql_query(query2, conn)
```

# WHERE Queries
This section filters table rows using the WHERE clause.

```python
# This query filters rows to show orders with Delivered status
query3 = "SELECT OrderID, CustomerID, Product, TotalPrice, OrderStatus FROM orders WHERE OrderStatus = 'Delivered' LIMIT 5;"
pd.read_sql_query(query3, conn)
```

```python
# This query filters orders with a total price greater than 2500
query4 = 'SELECT OrderID, Product, Quantity, TotalPrice FROM orders WHERE TotalPrice > 2500 LIMIT 5;'
pd.read_sql_query(query4, conn)
```

# ORDER BY
This section sorts query output based on total price.

```python
# This query sorts orders by TotalPrice in descending order to see highest sales
query5 = 'SELECT OrderID, Product, Quantity, TotalPrice FROM orders ORDER BY TotalPrice DESC LIMIT 5;'
pd.read_sql_query(query5, conn)
```

# GROUP BY
This section groups data by product category to count orders.

```python
# This query counts total orders for each product category using COUNT()
query6 = 'SELECT Product, COUNT(*) AS TotalOrders FROM orders GROUP BY Product ORDER BY TotalOrders DESC;'
pd.read_sql_query(query6, conn)
```

# Aggregate Functions
This section uses aggregate functions SUM() and AVG() with GROUP BY.

```python
# This query calculates total sales revenue for each payment method using SUM()
query7 = 'SELECT PaymentMethod, SUM(TotalPrice) AS TotalSales FROM orders GROUP BY PaymentMethod ORDER BY TotalSales DESC;'
pd.read_sql_query(query7, conn)
```

```python
# This query calculates average order total price for each product using AVG()
query8 = 'SELECT Product, AVG(TotalPrice) AS AveragePrice FROM orders GROUP BY Product ORDER BY AveragePrice DESC;'
pd.read_sql_query(query8, conn)
```

In this project, I loaded an e-commerce dataset into SQLite using Python and Pandas. I wrote simple SQL queries using SELECT, WHERE, ORDER BY, GROUP BY, and aggregate functions (COUNT, SUM, AVG) to analyze sales data.
